```python
import os, time, copy, random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

import torchvision
from torchvision import datasets, models, transforms

from sklearn.model_selection import train_test_split
```

```python
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

```python
BATCH_SIZE = 32
IMG_SIZE = 224

EPOCHS_FULL = 25
EPOCHS_HEAD = 20

LR_FULL = 1e-4
LR_HEAD = 1e-3

EARLY_STOPPING_PATIENCE = 5
SMALL_TRAIN_FRACTION = 0.2
```

```python
import kagglehub
path = kagglehub.dataset_download("mathewribeiro/spiderscorpioncochroach")
print("Dataset path:", path)
```

```python
def find_imagefolder_root(base_path):
    base = Path(base_path)
    for root in [base] + list(base.rglob("*")):
        if root.is_dir():
            subs = [p for p in root.iterdir() if p.is_dir()]
            if len(subs) >= 2:
                for sd in subs:
                    if any(f.suffix.lower() in [".jpg",".png",".jpeg"] for f in sd.iterdir()):
                        return str(root)
    return None

IMAGE_ROOT = find_imagefolder_root(path)
print("IMAGE_ROOT:", IMAGE_ROOT)
```

```python
tmp_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor()
])

full_dataset = datasets.ImageFolder(IMAGE_ROOT, transform=tmp_transform)
class_names = full_dataset.classes
num_classes = len(class_names)

labels = np.array([s[1] for s in full_dataset.samples])
idx = np.arange(len(labels))

train_idx, temp_idx, y_train, y_temp = train_test_split(idx, labels, test_size=0.2, stratify=labels, random_state=42)
val_idx, test_idx, y_val, y_test = train_test_split(temp_idx, y_temp, test_size=0.5, stratify=y_temp, random_state=42)
```

```python
data_transforms = {
    "train": transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(10),
        transforms.ToTensor(),
        transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
    ]),
    "val": transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
    ]),
    "test": transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
    ])
}

train_dataset = datasets.ImageFolder(IMAGE_ROOT, transform=data_transforms["train"])
val_dataset   = datasets.ImageFolder(IMAGE_ROOT, transform=data_transforms["val"])
test_dataset  = datasets.ImageFolder(IMAGE_ROOT, transform=data_transforms["test"])

train_loader = torch.utils.data.DataLoader(torch.utils.data.Subset(train_dataset, train_idx), batch_size=BATCH_SIZE, shuffle=True)
val_loader   = torch.utils.data.DataLoader(torch.utils.data.Subset(val_dataset, val_idx), batch_size=BATCH_SIZE, shuffle=False)
test_loader  = torch.utils.data.DataLoader(torch.utils.data.Subset(test_dataset, test_idx), batch_size=BATCH_SIZE, shuffle=False)
```

```python
small_train_idx, _ = train_test_split(
    train_idx,
    test_size=1-SMALL_TRAIN_FRACTION,
    stratify=labels[train_idx],
    random_state=42
)

small_train_loader = torch.utils.data.DataLoader(
    torch.utils.data.Subset(train_dataset, small_train_idx),
    batch_size=BATCH_SIZE,
    shuffle=True
)
```

```python
class EarlyStopping:
    def __init__(self, patience=5):
        self.best = 0
        self.counter = 0
        self.patience = patience
        self.best_state = None

    def step(self, val_acc, model):
        if val_acc > self.best:
            self.best = val_acc
            self.counter = 0
            self.best_state = copy.deepcopy(model.state_dict())
            return False
        else:
            self.counter += 1
            return self.counter >= self.patience
```

```python
@torch.no_grad()
def evaluate_accuracy(model, loader):
    model.eval()
    correct,total = 0,0
    for x,y in loader:
        x,y = x.to(device), y.to(device)
        preds = model(x).argmax(1)
        correct += (preds==y).sum().item()
        total += y.size(0)
    return 100*correct/total

def train_one_epoch(model, loader, optimizer, loss_fn):
    model.train()
    correct,total,loss_sum = 0,0,0
    for x,y in loader:
        x,y = x.to(device), y.to(device)
        optimizer.zero_grad()
        out = model(x)
        loss = loss_fn(out,y)
        loss.backward()
        optimizer.step()
        loss_sum += loss.item()*y.size(0)
        correct += (out.argmax(1)==y).sum().item()
        total += y.size(0)
    return loss_sum/total, 100*correct/total
```

```python
def build_model():
    m = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
    m.fc = nn.Linear(m.fc.in_features, num_classes)
    return m.to(device)
```

```python
def run_experiment(train_loader, mode_name, epochs, lr, freeze_backbone):
    model = build_model()

    if freeze_backbone:
        for p in model.parameters():
            p.requires_grad = False
        for p in model.fc.parameters():
            p.requires_grad = True
        optimizer = optim.Adam(model.fc.parameters(), lr=lr)
    else:
        optimizer = optim.Adam(model.parameters(), lr=lr)

    loss_fn = nn.CrossEntropyLoss()
    es = EarlyStopping(EARLY_STOPPING_PATIENCE)

    start = time.time()

    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, loss_fn)
        val_acc = evaluate_accuracy(model, val_loader)
        print(f"[{mode_name}] Epoch {epoch+1} | Train {train_acc:.2f}% | Val {val_acc:.2f}%")

        if es.step(val_acc, model):
            print(f"Early stopping {mode_name}")
            break

    elapsed = time.time() - start
    model.load_state_dict(es.best_state)
    test_acc = evaluate_accuracy(model, test_loader)

    return {
        "mode": mode_name,
        "val_acc": es.best,
        "test_acc": test_acc,
        "time_sec": elapsed
    }, model
```

```python
results = []

r, model_sf = run_experiment(small_train_loader, "Small + Full", 20, LR_FULL, freeze_backbone=False)
results.append(r)

r, model_sh = run_experiment(small_train_loader, "Small + Head", 20, LR_HEAD, freeze_backbone=True)
results.append(r)

r, model_lf = run_experiment(train_loader, "Large + Full", 25, LR_FULL, freeze_backbone=False)
results.append(r)

r, model_lh = run_experiment(train_loader, "Large + Head", 20, LR_HEAD, freeze_backbone=True)
results.append(r)
```

```python
test_subset = torch.utils.data.Subset(test_dataset, test_idx)

def predict_image(model, image_path, transform, class_names):
    img = Image.open(image_path).convert("RGB")
    x = transform(img).unsqueeze(0).to(device)
    with torch.no_grad():
        probs = torch.softmax(model(x),1)
    conf,pred = probs.max(1)
    return class_names[pred.item()], conf.item(), probs.cpu().numpy()[0]
```

```python
def compare_multiple(modelA, modelB, subset, n=8):
    indices = random.sample(range(len(subset)), n)
    plt.figure(figsize=(16,8))
    for i,idx in enumerate(indices):
        img_tensor,true_label = subset[idx]
        original_index = test_idx[idx]
        path = test_dataset.samples[original_index][0]
        true_class = class_names[true_label]

        pA,_,_ = predict_image(modelA, path, data_transforms["test"], class_names)
        pB,_,_ = predict_image(modelB, path, data_transforms["test"], class_names)

        ax = plt.subplot(2,4,i+1)
        ax.imshow(Image.open(path))
        ax.axis("off")
        ax.set_title(f"T:{true_class}\nF:{pA} | H:{pB}", fontsize=10)
    plt.show()

compare_multiple(model_lf, model_lh, test_subset, n=8)
```

# Conclusion

In this project, transfer learning was applied to an image classification task using a pretrained ResNet-18 model. The original classification head was replaced with a new fully connected layer adapted to the target dataset consisting of three insect classes: cockroach, scorpion, and spider. Two different transfer learning strategies were evaluated:

1. **Full fine-tuning**, where all layers of the pretrained network were retrained on the new dataset.
2. **Head-only training**, where the pretrained backbone was frozen and only the new classification layer was trained.

To further analyze the effect of dataset size, each strategy was applied to two versions of the dataset:
- A **small dataset** (20% of the training data),
- A **large dataset** (the full training set).

This resulted in four experimental settings: Small + Full, Small + Head, Large + Full, and Large + Head.

---

## Results Summary

| Setting        | Validation Accuracy | Test Accuracy | Training Time (s) |
|----------------|----------------------|---------------|-------------------|
| Small + Full   | 93%                  | 92%           | ~25               |
| Small + Head   | 87%                  | 83%           | ~21               |
| Large + Full   | 99%                  | 96%           | ~73               |
| Large + Head   | 93%                  | 91%           | ~70               |

---

## Analysis

The **full fine-tuning approach consistently achieved higher accuracy** than the head-only approach in both dataset sizes. This is because retraining all layers allows the network to adapt its low-level and high-level feature representations to the specific characteristics of the insect dataset.

However, full fine-tuning also required **more training time**, especially for the larger dataset. In contrast, the head-only approach was slightly faster and more computationally efficient, but it achieved lower accuracy because it relied entirely on the pretrained ImageNet features.

When using the **small dataset**, the performance gap between the two methods was smaller, and both models performed reasonably well. When using the **large dataset**, the advantage of full fine-tuning became more evident, as the model benefited from having more data to safely update all parameters without overfitting.

---

## Conclusion

These experiments demonstrate that:

- Full fine-tuning provides the best performance when sufficient data and computational resources are available.
- Freezing the backbone and training only the classifier head is a faster and more lightweight alternative, suitable when computational resources are limited or when the dataset is very small.
- Dataset size plays a critical role in determining which transfer learning strategy is more effective.

Overall, the project shows how different transfer learning strategies and dataset sizes affect both model performance and training efficiency, highlighting the trade-off between accuracy and computational cost.
